## Missing Values

```python
missing = df.isnull().sum()
print('Missing values per column:')
print(missing[missing > 0] if missing.any() else 'None — dataset is complete.')
```

## Descriptive Statistics

```python
df.describe().T
```

## Class Distribution

```python
fraud = df['Class'].sum()
legit = len(df) - fraud
print(f'Total transactions : {len(df):,}')
print(f'Fraud (Class=1)    : {fraud:,}  ({fraud/len(df):.4%})')
print(f'Legit (Class=0)    : {legit:,}  ({legit/len(df):.4%})')
```

## Key Takeaways

- The dataset is **highly imbalanced** — fraud is ~0.17% of all transactions.
- `V1`–`V28` are PCA-transformed (already zero-mean, unit-variance).
- `Time` and `Amount` are raw and will need scaling.
- No missing values — no imputation required.
